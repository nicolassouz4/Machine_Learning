## Import Libraries
**Import pandas,seaborn, and the usual libraries.**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
%matplotlib inline
```

## Get the Data
** Reading the 'KNN_Project_Data csv file into a dataframe **

```python
df = pd.read_csv('KNN_Project_Data')
df.head()
```

# EDA

Since this data is artificial, we'll just do a large pairplot with seaborn.

**Using seaborn on the dataframe to create a pairplot with the hue indicated by the TARGET CLASS column.**

```python
sns.pairplot(data=df,hue='TARGET CLASS')
```

# Standardize the Variables

Standardize the variables

**Import StandardScaler from Scikit learn.**

```python
from sklearn.preprocessing import StandardScaler
```

**Creating a StandardScaler() object called scaler.**

```python
scaler = StandardScaler()
```

**Fit scaler to the features.**

```python
scaler.fit(df.drop('TARGET CLASS',axis=1))
```

**Using the .transform() method to transform the features to a scaled version.**

```python
scaled_features = scaler.transform(df.drop('TARGET CLASS',axis=1))
```

```python
df_ft = pd.DataFrame(scaled_features, columns = df.columns[:-1])
df_ft.head()
```

# Train Test Split

**Usin train_test_split to split data into a training set and a testing set.**

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(scaled_features, df['TARGET CLASS'], test_size=0.3)
```

# Using KNN

**Import KNeighborsClassifier from scikit learn.**

```python
from sklearn.neighbors import KNeighborsClassifier
```

**Creating a KNN model instance with n_neighbors=1**

```python
knn = KNeighborsClassifier(n_neighbors=1)
```

**Fit this KNN model to the training data.**

```python
knn.fit(X_train, y_train)
```

# Predictions and Evaluations
Evaluate KNN model!

**Using the predict method to predict values using KNN model and X_test.**

```python
pred = knn.predict(X_test)
```

** Create a confusion matrix and classification report.**

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(confusion_matrix(y_test, pred))
```

```python
print(classification_report(y_test, pred))
```

# Choosing a K Value
Using the elbow method to pick a good K Value!

**Creating a for loop that trains various KNN models with different k values, then keep track of the error_rate for each of these models with a list.**

```python
error_rate = []

for i in range(1,50):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train, y_train)
    pred_i = knn.predict(X_test)
    error_rate.append(np.mean(pred_i != y_test))
```

**Creating plots using the information from the for loop.**

```python
plt.figure(figsize=(10,6))
plt.plot(range(1,50), error_rate, color='blue', linestyle='dashed', marker='o',
         markerfacecolor='red', markersize=10)
plt.title('Error Rate vs. K Value')
plt.xlabel('K')
plt.ylabel('Error Rate')
```

## Retrain with new K Value

**Retraining model with the best K value and re-do the classification report and the confusion matrix.**

```python
knn_30 = KNeighborsClassifier(n_neighbors=30)
knn_30.fit(X_train, y_train)
pred_30= knn_30.predict(X_test)
error_rate.append(np.mean(pred_30 != y_test))
```

```python
print(confusion_matrix(y_test, pred_30))
print(classification_report(y_test, pred_30))
```
